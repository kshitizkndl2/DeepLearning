```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import make_moons
x, y = make_moons(noise = 3, n_samples=200, random_state=42)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(x=x[:,0],y=x[:,1],hue=y)
plt.show()

```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
import tensorflow 
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adagrad,RMSprop,Adam
```

```python
adagrad = Adagrad(
    learning_rate=0.001,
    initial_accumulator_value=0.1,
    epsilon=1e-07,
    weight_decay=None,
    clipnorm=None,
    clipvalue=None,
    global_clipnorm=None,
    use_ema=False,
    ema_momentum=0.99,
    ema_overwrite_frequency=None,
    name="Adagrad",
)
```

```python
model1 = Sequential([
    Dense(64,activation='relu',input_shape = (2,)),
     Dense(32,activation = 'relu'),
     Dense(1,activation = 'linear')
])
```

```python
model1.compile(loss='binary_crossentropy', optimizer = adagrad, metrics = ['accuracy'])
```

```python
history1 = model1.fit(x_train,y_train, epochs = 200,validation_data = (x_test,y_test) )
```

```python
from mlxtend.plotting import plot_decision_regions
plt.title('AdaGrad')
plot_decision_regions(x,y,model1)
```

```python
rmsprop = RMSprop(
    learning_rate=0.001,
    rho=0.9,
    momentum=0.0,
    epsilon=1e-07,
    centered=False,
    weight_decay=None,
    clipnorm=None,
    clipvalue=None,
    global_clipnorm=None,
    use_ema=False,
    ema_momentum=0.99,
    ema_overwrite_frequency=None,
    loss_scale_factor=None,
    gradient_accumulation_steps=None,
    name=None,
   
)
```

```python
model2 = Sequential([
    Dense(64,activation='relu',input_shape = (2,)),
     Dense(32,activation = 'relu'),
     Dense(1,activation = 'linear')
])
```

```python
model2.compile(loss='binary_crossentropy', optimizer = rmsprop, metrics = ['accuracy'])
```

```python
history2 = model2.fit(x_train,y_train, epochs = 200,validation_data = (x_test,y_test) )
```

```python
plt.title('RMSProp')
plot_decision_regions(x,y,model2)
```

```python
adam = Adam(
    learning_rate=0.001,
    beta_1=0.9,
    beta_2=0.999,
    epsilon=1e-07,
    amsgrad=False,
    weight_decay=None,
    clipnorm=None,
    clipvalue=None,
    global_clipnorm=None,
    use_ema=False,
    ema_momentum=0.99,
    ema_overwrite_frequency=None,
    loss_scale_factor=None,
    gradient_accumulation_steps=None,
    name=None
)
```

```python
model3 = Sequential([
    Dense(64,activation='relu',input_shape = (2,)),
     Dense(32,activation = 'relu'),
     Dense(1,activation = 'linear')
])
```

```python
model3.compile(loss='binary_crossentropy', optimizer = adam, metrics = ['accuracy'])
```

```python
history3 = model3.fit(x_train,y_train, epochs = 200,validation_data = (x_test,y_test) )
```

```python
plt.title('adam')
plot_decision_regions(x,y,model3)
```
